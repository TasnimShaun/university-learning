# Task-01: Sequential Search
#Write a program that stores 10 integers in an array (or Python list), asks the user for a target number, andchecks the elements one by one using a loop. Print “Element Found” with its index, or “Element Not Found”.
`Example array: 12, 5, 8, 20, 15, 7, 30, 10, 25, 18
Example input: 20 Expected output: Element Found! Index: 3`

```python
# use function
def Sequential_Search(list_, target):
    for i in range(len(list_)):
        if list_[i] == target:
            return i
    return -1

def main():
    list_ = [12, 5, 8, 20, 15, 7, 30, 10, 25, 18]


    target = int(input("Enter the target number to search: "))

    index = Sequential_Search(list_, target)

    if index != -1:
        print(f"Element Found! Index: {index}")
    else:
        print("Element Not Found")

if __name__ == "__main__":
    main()
```

# Task-02: Binary Search
#Write a program that uses a sorted array of 10 integers. Ask for a target, calculate the middle index using lowand high boundaries, and continue searching in the left or right half until the target is found or the searchends. Print the index or “Element Not Found”.

`Example sorted array: 2, 5, 8, 12, 15, 20, 25, 30, 35, 40`

`Example input: 25 Expected output: Element Found! Index: 6`

`Important: Binary search requires sorted input. Do not apply it directly to an unsorted array.`


```python
# use function

def Binary_Search(list_, target):
  left = 0
  right = len(list_) - 1
  while( left <= right):
    mid = left +( right - left) // 2
    if list_[mid] == target:
      return mid
    elif list_[mid] < target:
      left = mid + 1
    else:
      right = mid - 1
    return -1

def main():
  list_ = [5, 10, 15, 20, 25, 30, 35, 40]
  target = int(input("Enter target number: "))
  index = Binary_Search(list_, target)

  if index != -1:
        print(f"Element Found! Index: {index}")
  else:
        print("Element Not Found")

if __name__ == "__main__":
    main()
```

```python

list_ = [5, 10, 15, 20, 25, 30, 35, 40]

target = int(input("Enter target number: "))

pos = -1
left = 0
right = len(list_) - 1

while left <= right:
    mid = left + (right - left) // 2

    if list_[mid] == target:
        pos = mid
        break
    elif list_[mid] < target:
        left = mid + 1
    else:
        right = mid - 1


if pos != -1:
    print("Element Found! Index:",pos)
else:
    print("Element Not Found")
```

#Task-03: Manual tracing


##Given Array: [5, 10, 15, 20, 25, 30, 35, 40]

Target: 15

| Step | Sequential Search | Binary Search |
|---|---|---|
| 1 | 5 (Index 0) | 20 (Index 3) |
| 2 | 10 (Index 1) | 5 (Index 0) |
| 3 | 15 (Index 2, Found) | 10 (Index 1) |
| 4 | — | 15 (Index 2, Found) |

##Sequential search checks each element from the beginning until it finds 15. It checks 5, 10, and 15, so it requires 3 comparisons.

##Binary search checks the middle element, 20, first. Since 15 is smaller than 20, it searches the left half. Then it checks 5, followed by 10 and finally 15. It requires 4 comparisons.

##In this example, sequential search is faster because it finds the target in 3 comparisons, while binary search needs 4 comparisons.
