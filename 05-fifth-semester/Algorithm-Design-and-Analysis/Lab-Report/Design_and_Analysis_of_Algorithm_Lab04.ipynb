#Task 2 – Understand Recursion

**Iteration:** `Repetition of a block of code using a loop construct until a condition is met.`

**Recursion:** `A technique where a function calls itself with a simpler input until it reaches a base case.`

**Base case:** `The condition that stops the recursive calls and returns a value directly without further recursion.`

#### Labeled example (factorial)

- **Base case:** `F(0) = 1` → when n = 0, return 1 directly (no recursive call).
- **Recursive rule:** `F(n) = n × F(n−1)` for n > 0 → the function calls itself with n−1.Task 2 – Understand Recursion


#Task 03 - Factorial using iteration and recursion


## Pseudocode (iterative)
---

function factorialIter(n):

    if n < 0: return ERROR
    result ← 1
    for i from 1 to n:
        result ← result * i
    return result


---
## Pseudocode (recursive)

function factorialRec(n):

    if n < 0: return ERROR
    if n == 0 or n == 1: return 1        // base case
    return n * factorialRec(n - 1)        // recursive call



```python
# ============================================
# Task 3: Factorial - Iterative vs Recursive
# ============================================

def factorial_iter(n):
    """Iterative factorial using a loop."""
    if n < 0:
        raise ValueError("Negative input not allowed")
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


def factorial_rec(n):
    """Recursive factorial using base case n = 0 or 1."""
    if n < 0:
        raise ValueError("Negative input not allowed")
    if n == 0 or n == 1:          # base case
        return 1
    return n * factorial_rec(n - 1)   # recursive call


# ---- Test inputs required by the lab ----
test_inputs = [0, 1, 5, 7]

print(f"{'n':>3} | {'Iterative':>12} | {'Recursive':>12} | {'Match?':>6}")
print("-" * 45)

for n in test_inputs:
    it = factorial_iter(n)
    rc = factorial_rec(n)
    match = "YES" if it == rc else "NO"
    print(f"{n:>3} | {it:>12} | {rc:>12} | {match:>6}")

# ---- Negative input handling demo ----
print("\n--- Negative input test ---")
try:
    factorial_iter(-1)
except ValueError as e:
    print("Iterative -1 ->", e)

try:
    factorial_rec(-1)
except ValueError as e:
    print("Recursive -1 ->", e)
```

#Task 4 – Fibonacci:
#Pseudocode + Code
Definition: F(0)=0, F(1)=1, F(n)=F(n−1)+F(n−2) for n ≥ 2.

Pseudocode (iterative, returns F(n))


---


function fibIter(n):

    if n < 0: return ERROR
    if n == 0: return 0
    if n == 1: return 1
    prev2 ← 0      // F(0)
    prev1 ← 1      // F(1)
    for i from 2 to n:
        curr ← prev1 + prev2
        prev2 ← prev1
        prev1 ← curr
    return prev1

Pseudocode (naive recursive)

---
function fibRec(n):

    if n < 0: return ERROR
    if n == 0: return 0        // base case 1
    if n == 1: return 1        // base case 2
    return fibRec(n - 1) + fibRec(n - 2)   // two recursive calls


```python


def fib_iter(n):
    """Iterative Fibonacci returning F(n) using two variables."""
    if n < 0:
        raise ValueError("Negative input not allowed")
    if n == 0:
        return 0
    if n == 1:
        return 1
    prev2, prev1 = 0, 1            # F(0), F(1)
    for i in range(2, n + 1):
        curr = prev1 + prev2
        prev2, prev1 = prev1, curr
    return prev1


def fib_rec(n):
    """Naive recursive Fibonacci with two base cases."""
    if n < 0:
        raise ValueError("Negative input not allowed")
    if n == 0:                     # base case 1
        return 0
    if n == 1:                     # base case 2
        return 1
    return fib_rec(n - 1) + fib_rec(n - 2)   # two recursive calls

def first_terms(func, n):
    return [func(i) for i in range(n)]


test_ns = [5, 8, 10]

for n in test_ns:
    seq_iter = first_terms(fib_iter, n)
    seq_rec  = first_terms(fib_rec, n)
    match = "YES" if seq_iter == seq_rec else "NO"
    print(f"n = {n}")
    print(f"  Iterative : {seq_iter}")
    print(f"  Recursive : {seq_rec}")
    print(f"  Match?    : {match}\n")

# ---- Single F(n) values side by side ----
print(f"{'n':>3} | {'F_iter(n)':>10} | {'F_rec(n)':>10} | {'Match?':>6}")
print("-" * 42)
for n in [0, 1, 2, 5, 8, 10]:
    a, b = fib_iter(n), fib_rec(n)
    print(f"{n:>3} | {a:>10} | {b:>10} | {'YES' if a==b else 'NO':>6}")

# ---- Negative input check ----
print("\n--- Negative input test ---")
for f in (fib_iter, fib_rec):
    try:
        f(-1)
    except ValueError as e:
        print(f"{f.__name__}(-1) ->", e)

```

```python


def build_tree(n, prefix="", is_last=True, is_root=True):
    """Recursively build and print the call tree."""
    # branch drawing
    if is_root:
        connector = ""
    else:
        connector = "└── " if is_last else "├── "

    # base case → show returned value
    if n < 2:
        print(f"{prefix}{connector}fib({n}) → {n}")
        return

    # recursive node
    print(f"{prefix}{connector}fib({n})")

    # child prefix
    if is_root:
        child_prefix = ""
    else:
        child_prefix = prefix + ("    " if is_last else "│   ")

    build_tree(n - 1, child_prefix, False, False)
    build_tree(n - 2, child_prefix, True,  False)


print("=== Task 5B: fibonacci(5) call tree ===\n")
build_tree(5)
```

```python
from collections import Counter

call_count = Counter()

def fib_trace(n):
    call_count[n] += 1
    if n == 0: return 0
    if n == 1: return 1
    return fib_trace(n-1) + fib_trace(n-2)

val = fib_trace(5)

print("=== Task 5B: fibonacci(5) call counts ===\n")
print(f"{'Call':>10} | {'Times called':>12}")
print("-" * 27)
for k in sorted(call_count, reverse=True):
    print(f"{'fib('+str(k)+')':>10} | {call_count[k]:>12}")
print("-" * 27)
print(f"{'TOTAL':>10} | {sum(call_count.values()):>12}")
print(f"\nfib(5) = {val}")
```

```python
# Count invocations for factorial(4) and fibonacci(5)
fac_calls = 0
def fac(n):
    global fac_calls
    fac_calls += 1
    if n <= 1: return 1
    return n * fac(n-1)

fib_calls = 0
def fib(n):
    global fib_calls
    fib_calls += 1
    if n < 2: return n
    return fib(n-1) + fib(n-2)

fac(4); print(f"factorial(4) → total invocations = {fac_calls}")
fib(5); print(f"fibonacci(5) → total invocations = {fib_calls}")
```

### Task 5D – Complexity Comparison

| Implementation | Time complexity | Auxiliary space |
|---|---|---|
| Factorial, iterative | **O(n)** | **O(1)** |
| Factorial, recursive | **O(n)** | **O(n)** (call stack, n+1 frames) |
| Fibonacci, iterative (F(n)) | **O(n)** | **O(1)** (two variables) |
| Fibonacci, naive recursive | **O(2ⁿ)** (≈ φⁿ, φ≈1.618) | **O(n)** (max recursion depth) |

**Notes:**
- Iterative factorial: n multiplications, one accumulator → O(n) time, O(1) space.
- Recursive factorial: n calls, each O(1); stack holds n+1 frames → O(n) time, O(n) space.
- Iterative Fibonacci: loop 2..n, only prev1/prev2 stored → O(n) time, O(1) space.
- Naive recursive Fibonacci: T(n)=T(n−1)+T(n−2)+O(1) → exponential; depth n → O(n) stack.

```python
# ============================================
# Task 6: Test and Compare
# ============================================

# ---------- Factorial ----------
def factorial_iter(n):
    if n < 0: raise ValueError("negative input")
    r = 1
    for i in range(1, n + 1): r *= i
    return r

def factorial_rec(n):
    if n < 0: raise ValueError("negative input")
    if n <= 1: return 1
    return n * factorial_rec(n - 1)

# ---------- Fibonacci ----------
def fib_iter(n):
    if n < 0: raise ValueError("negative input")
    if n == 0: return 0
    if n == 1: return 1
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

def fib_rec(n):
    if n < 0: raise ValueError("negative input")
    if n < 2: return n
    return fib_rec(n - 1) + fib_rec(n - 2)


# ============================================
# Step A: Factorial — iterative vs recursive
# ============================================
print("=" * 50)
print("STEP A: Factorial test")
print("=" * 50)
print(f"{'n':>3} | {'Iterative':>10} | {'Recursive':>10} | {'Equal?':>6}")
print("-" * 40)
all_fac_ok = True
for n in [0, 1, 5, 7]:
    it, rc = factorial_iter(n), factorial_rec(n)
    ok = it == rc
    all_fac_ok &= ok
    print(f"{n:>3} | {it:>10} | {rc:>10} | {'YES' if ok else 'NO':>6}")
print(f"\nAll factorial results equal? {'YES ✅' if all_fac_ok else 'NO ❌'}")


# ============================================
# Step B: Fibonacci — iterative vs recursive
# ============================================
print("\n" + "=" * 50)
print("STEP B: Fibonacci test (first n terms)")
print("=" * 50)
all_fib_ok = True
for n in [5, 8, 10]:
    seq_it  = [fib_iter(i) for i in range(n)]
    seq_rec = [fib_rec(i)  for i in range(n)]
    ok = seq_it == seq_rec
    all_fib_ok &= ok
    print(f"n = {n}")
    print(f"  Iterative : {seq_it}")
    print(f"  Recursive : {seq_rec}")
    print(f"  Equal?    : {'YES ✅' if ok else 'NO ❌'}\n")

# single F(n) check
print(f"{'n':>3} | {'F_iter':>7} | {'F_rec':>7} | {'Equal?':>6}")
print("-" * 32)
for n in [5, 8, 10]:
    a, b = fib_iter(n), fib_rec(n)
    print(f"{n:>3} | {a:>7} | {b:>7} | {'YES' if a==b else 'NO':>6}")

print(f"\nAll Fibonacci sequences equal? {'YES ✅' if all_fib_ok else 'NO ❌'}")


# ============================================
# Step D: Short answers
# ============================================
print("\n" + "=" * 50)
print("STEP D: Short answers")
print("=" * 50)
print("""
1) Why does recursion require a base case?
   → Without a base case, the function calls itself forever
     (infinite recursion) until the call stack overflows and the
     program crashes. The base case gives a stopping condition so
     the recursive calls can unwind and return a real value.

2) Which Fibonacci method repeats calculations?
   → The NAIVE RECURSIVE version. It recomputes the same
     subproblems many times. For fib(5), fib(3) is computed 2×,
     fib(2) 3×, fib(1) 5× — giving exponential O(2ⁿ) work.

3) Which Fibonacci method would you choose for large n, and why?
   → The ITERATIVE version: O(n) time, O(1) space. It computes
     each value only once, uses just two variables, avoids stack
     overflow, and stays fast for large n. (Naive recursion is
     exponential and impractical beyond n ≈ 40.)
""")
```
