## Swap two numbers

```python

x = int(input("Please enter x: "))
y = int(input("Please enter y: "))

temp = x
x = y
y = temp

print(f"After swapping: x = {x} and y = {y}")
```

```python
# another method

x = int(input("Please enter x: "))
y = int(input("Please enter y: "))

x, y = y, x

print(f"After swapping: x = {x} and y = {y}")
```

```python
# if / else if / else: Grades A to F

for i in range (1,8): # for 7 testcase
  marks = int(input("please enter your marks: "))
  if marks < 0 or marks > 100:
    print("Invalid")
  elif marks >= 90:
    print("A")
  elif marks >= 80:
    print("B")
  elif marks >= 70:
    print("C")
  elif marks >= 60:
    print("D")
  elif marks >= 50:
    print("E")
  else:
    print("F")
```
